```python
from pydantic import BaseModel

class PydanticUser(BaseModel):
    name : str

user2 = PydanticUser(name="문자열")
print('Pydantic 결과:', user2.name)
print('Pydantic 데이터 타입:', type(user2.name))
```

```python
from pydantic import BaseModel

# 1. 데이터 모델 정의
class UserRegister(BaseModel):
    user_id: int
    username: str
    email: str
    is_admin: bool = False

# 2. 올바른 데이터 입력과 객체 생성
# user = UserRegister(
#     user_id=1,
#     username='홍길동',
#     email='hong@example.com'
# )

input_data = {
    "user_id": "1004", # 문자열로 들어왔지만 int로 자동 변환(Parsing)됩니다.
    "username": "codemaster",
    "email": "code@example.com"
    # is_admin은 생략되었으므로 기본값 False가 적용됩니다.
}

user1 = UserRegister(**input_data) # 언팩킹
user1.user_id
```

```python
user1.model_dump()
```

```python
user1.model_dump_json()
```
