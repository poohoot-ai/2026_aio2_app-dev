```python
import urllib.request
import urllib.error
import datetime # 예제용

# 예제 URL (안정적인 사이트 사용 권장)
target_url = "https://www.python.org/"
# target_url = "http://httpbin.org/get" # HTTP 테스트용 사이트

print(f"--- URL GET 요청 시작: {target_url} ---")
try:
    # with 문을 사용하여 자동으로 리소스 정리
    with urllib.request.urlopen(target_url, timeout=10) as response:
        status = response.status
        print(f"상태 코드: {status}")

        headers = response.headers
        print(f"콘텐츠 타입: {headers.get('Content-Type', '알 수 없음')}")

        # 헤더에서 인코딩 정보 얻기 (없으면 utf-8 가정)
        encoding = headers.get_content_charset() or 'utf-8'
        print(f"감지된/가정된 인코딩: {encoding}")

        if 200 <= status < 300: # 성공적인 응답 확인
            html_bytes = response.read() # 본문을 바이트로 읽기
            # 바이트를 문자열로 디코딩
            html_content = html_bytes.decode(encoding)
            print("\n--- 응답 내용 (일부) ---")
            print(html_content[:500] + "...") # 앞 500자만 출력
        else:
            print(f"오류 응답 수신: 상태 코드 {status}")

except urllib.error.HTTPError as e:
    # HTTP 오류 처리 (예: 404, 500)
    print(f"HTTP 오류 발생: {e.code} {e.reason}")
except urllib.error.URLError as e:
    # URL 관련 오류 처리 (예: 호스트 찾을 수 없음, 네트워크 연결 불가)
    print(f"URL 오류 발생: {e.reason}")
except TimeoutError:
    print("요청 시간 초과")
except Exception as e:
    print(f"기타 오류 발생: {e}")
```

```python
only_url = "https://api.open-meteo.com/v1/forecast?"
params_value = {
    "latitude":37.4022,
    "longitude":126.9764,
    "current":"temperature_2m,relative_humidity_2m",
    "timezone":"Asia/Seoul",
}
```

```python
import requests
response = requests.get(only_url, params=params_value, timeout=5.0)
response.json()
```

```python
# httpx

import httpx

async def get_weather():

    async with httpx.AsyncClient(timeout=5) as client:
        response = await client.get(
            only_url, 
            params=params_value
        )
        data = response.json()
    return data
```

```python
result = await get_weather()
print(result)
```
